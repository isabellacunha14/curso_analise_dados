#Soma Ingredientes



```python
#importacões
import sys
```

```python
#Declaração de Variaveis
preco_cenoura = 4.5
preco_acucar = 5
preco_ovos = 11
```

```python
#Funcao
def soma_ingredientes(tem_cenoura, tem_acucar, tem_ovos):
	total_compra=0
	if tem_cenoura:
		total_compra = total_compra + preco_cenoura
	if tem_acucar:
		total_compra = total_compra + preco_acucar
	if tem_ovos:
		total_compra = total_compra + preco_ovos
	return total_compra

total= soma_ingredientes(True,True,True)
print(total)
```

```python
terminal_tem_cenoura = True
terminal_tem_acucar = True
terminal_tem_ovos = True

total= soma_ingredientes(terminal_tem_cenoura,terminal_tem_acucar, terminal_tem_ovos)

print("Total: ", total)
```
